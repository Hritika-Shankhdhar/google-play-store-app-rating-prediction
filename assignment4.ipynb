Data Cleaning and Preprocessing

```python
#Implementing the data
data = pd.read_csv('C:\\Users\\KIIT\\OneDrive\\Desktop\\python\\datasets\\googleplaystore.csv')
```

```python
#Checking for any null values
pd.isnull(data).sum()
```

```python
#Droping all the null values 
data.dropna(subset=['Rating'], inplace=True)
data.drop(columns=['Current Ver'], inplace=True)

#Filling some columns having null values
data['Type'].fillna(data['Type'].mode()[0], inplace=True)
data['Content_Rating'].fillna(data['Content_Rating'].mode()[0], inplace=True)
data['Android Ver'].fillna(data['Android Ver'].mode()[0], inplace=True)
```

```python
pd.isnull(data).sum()
```

```python
#Finding if there are any duplicaye values
print('No. of duplicate rows:', data.duplicated().sum())
```

```python
#Remove duplicate values
data.drop_duplicates(inplace = True)
```

```python
print(data.dtypes)
```

```python
data.head()
```

```python
#Getting the unique datasets from columns containing datatype as object 
for col in data.columns:
    if data[col].dtype == 'object':
        print(f"\nSample values in '{col}':")
        print(data[col].unique()[:5])
```

```python
#Converting object datatypes to numeric
# Fixing Reviews
data['Reviews'] = pd.to_numeric(data['Reviews'], errors = 'coerce')
data['Reviews'] = data['Reviews'].fillna(data['Reviews'].median())

#Fixing Installs
data['Installs'] = data['Installs'].astype(str).str.replace('[+,]', '', regex=True)
data['Installs'] = pd.to_numeric(data['Installs'], errors='coerce')
data['Installs']= data['Installs'].fillna(data['Installs'].median())

#Fixing Price
data['Price'] = data['Price'].astype(str).str.replace('$', '', regex=True)
data['Price'] = pd.to_numeric(data['Price'], errors='coerce')
data['Price']= data['Price'].fillna(data['Price'].median())

#Fixing Size
def convert_size(size):
    if isinstance(size, str):
        if 'M' in size:
            return float(size.replace('M', '').replace(',', '.')) * 1e6
        elif 'k' in size:
            return float(size.replace('k', '').replace(',', '.')) * 1e3
        elif size == 'Varies with device':
            return np.nan
    return np.nan
data['Size'] = data['Size'].apply(convert_size)
data['Installs'] = pd.to_numeric(data['Installs'], errors='coerce')
data['Size'].fillna(data['Size'].median(), inplace=True)

#Fixing last updated to date
data['Last Updated'] = pd.to_datetime(data['Last Updated'], errors='coerce')
data['Days Since Update'] = (pd.Timestamp.now() - data['Last Updated']).dt.days
# Drop the original datetime column after converting
data.drop(columns=['Last Updated'], inplace=True)
# Drop rows with missing days (caused by invalid 'Last Updated')
data = data.dropna(subset=['Days Since Update'])

#Fixing android version
data['Android Ver'] = data['Android Ver'].astype(str).str.extract(r'(\d+\.\d+)', expand=False)
data['Android Ver'] = pd.to_numeric(data['Android Ver'], errors='coerce')
data['Android Ver'].fillna(data['Android Ver'].median(), inplace=True)
```

```python
#Encodng the Categorical Features using Label and OHE
#Cleaning the Categorical data
data_clean = data.copy()
data_clean['Genres'] = data_clean['Genres'].astype(str).str.split(';').str[0]
data_clean['Type'] = data_clean['Type'].replace('0', 'Free')

#Categorical coulmns
categorical_cols = ['Category', 'Type', 'Content_Rating', 'Genres']

for col in categorical_cols:
    if data_clean[col].isna().all():
        print(f"Column{col} is all NaNs. Filling with placeholder")
        data_clean[col] = 'Unknown'
    else:
        mode_val = data_clean[col].mode()[0]
        data_clean[col] = data_clean[col].fillna(mode_val)

#Label Encodeing (for tree models)
from sklearn.preprocessing import LabelEncoder
data_label = data_clean.copy()
le = LabelEncoder()
for col in categorical_cols:
    data_label[col] = le.fit_transform(data_label[col].astype(str))

#One Hot Encoding (for Linear/ KNN Models)
from sklearn.preprocessing import OneHotEncoder
data_ohe = data_clean.copy()
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
if data_ohe[categorical_cols].dropna().shape[0] == 0:
    print("No rows available for One Hot Encoder after Encoder")
else:
    encoded_array = ohe.fit_transform(data_ohe[categorical_cols])
    encoded_df = pd.DataFrame(
    encoded_array, 
    columns = ohe.get_feature_names_out(categorical_cols),
    index = data_ohe.index)


    data_ohe.drop(columns =categorical_cols, inplace = True)
    data_ohe = pd.concat([data_ohe, encoded_df], axis=1)

    #Final checking
    print(data_label.shape)
    print(data_ohe.shape)
```

Feature Engineering

```python
#Defining target variable clearly (e.g., "Installs" > 1,000,000 as success).
data_label['Success'] = data_label['Installs'].apply(lambda x:1 if x> 1_000_000 else 0)
data_ohe['Success'] = data_label['Installs'].apply(lambda x:1 if x>1_000_000 else 0) 
```

```python
#Selecting features relevant to the prediction task.
#For Label encoding
features = ['Rating', 'Reviews', 'Size', 'Price', 'Category', 'Type', 'Content_Rating', 'Genres', 'Days Since Update']
x_label = data_label[features]
y_label = data_ohe['Success']
```

<span style="font-size:30px">
Question = Justify your feature selection process.  
<span style="font-size:20px">

**Rating** =  higher-rated apps are more likely to be successful.  
**Size** = Can influence download behavior depending on the user’s device  
**Reviews** = strong signal of user engagement and popularity.
**Price** =Determines if the app is free or paid  
**Days Since Update** = Measures recency of updates; frequently updated apps tend to perform better.  
**Category** = Different app categories show varied install trends and competition levels.  
**Type** = Indicates if the app is free or paid, free apps get high users.  
**Genres** = Provides finer granularity than Category, helping capture app theme better.


```python
#For OHE
x_ohe = data_ohe.drop(columns = ['App', 'Installs', 'Success'])
y_ohe = data_ohe['Success']
```

```python
#Spliting the data into training and test sets (80/20 split).
from sklearn.model_selection import train_test_split

#spliting the Label Encoding version
x_train_label, x_test_label, y_train_label, y_test_label = train_test_split(x_label, y_label, test_size = 0.2, random_state = 42)

#spliting the Hot End Encode
x_train_ohe, x_test_ohe, y_train_ohe, y_test_ohe = train_test_split(x_ohe, y_ohe, test_size = 0.2, random_state = 42)
```

```python
#standardize the data
from sklearn.preprocessing import StandardScaler
numeric_cols = ['Rating', 'Reviews', 'Size', 'Price', 'Days Since Update']

#Scale label-encoded version
scaler_label = StandardScaler()
x_train_label[numeric_cols] = scaler_label.fit_transform(x_train_label[numeric_cols])
x_test_label[numeric_cols] = scaler_label.transform(x_test_label[numeric_cols])

#One hot encoded version
scaler_ohe = StandardScaler()
x_train_ohe[numeric_cols] = scaler_ohe.fit_transform(x_train_ohe[numeric_cols])
x_test_ohe[numeric_cols] = scaler_ohe.transform(x_test_ohe[numeric_cols])
```

```python
#evaluation function
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, roc_curve 
def evaluate_model (model_name, actual_labels, predicted_labels, predicted_probabilities):
    print(f"\nModel: {model_name}")
    print("Accuracy:", accuracy_score(actual_labels, predicted_labels))
    print("Precision:", precision_score(actual_labels, predicted_labels))
    print("Recall:", recall_score(actual_labels, predicted_labels))
    print("F1-score:", f1_score(actual_labels, predicted_labels))
    
          
```

<span style="font-size:30px">
Model Building

```python
#Logistic Regression
from sklearn.linear_model import LogisticRegression
Logistic = LogisticRegression(max_iter = 1000)

#For the OHE version
Logistic.fit(x_train_ohe, y_train_ohe)
y_predict_Logistic = Logistic.predict(x_test_ohe)
y_probability_Logistic = Logistic.predict_proba(x_test_ohe)[:,1]
evaluate_model('Logistic Regression (OHE)', y_test_ohe, y_predict_Logistic, y_probability_Logistic)
```

```python
#SVM
from sklearn.svm import SVC
svm = SVC(probability= True)

#For the OHE version
svm.fit(x_train_ohe, y_train_ohe)
y_predict_svm = svm.predict(x_test_ohe)
y_probability_svm = svm.predict_proba(x_test_ohe) [:,1]
evaluate_model("SVM (OHE)", y_test_ohe, y_predict_svm, y_probability_svm)
```

```python
#RandomForest
from sklearn.ensemble import RandomForestClassifier

#For the Label version
random_forest = RandomForestClassifier(random_state= 42)
random_forest.fit(x_train_label, y_train_label)
y_predict_random_forest = random_forest.predict(x_test_label)
y_probability_random_forest = random_forest.predict_proba(x_test_label)[:,1]
evaluate_model("Random Forest (Label)", y_test_label, y_predict_random_forest, y_probability_random_forest)
```

```python
#xgBoost
from xgboost import XGBClassifier
xgb_label = XGBClassifier(use_label_encoder=False, eval_metric='logloss')
xgb_label.fit(x_train_label, y_train_label)

y_predict_xgb_label = xgb_label.predict(x_test_label)
y_probability_xgb_label = xgb_label.predict_proba(x_test_label)[:, 1]

evaluate_model("XGBoost (Label)", y_test_label, y_predict_xgb_label, y_probability_xgb_label)
```

<span style="font-size:30px">
Model Evaluation

```python
#Calculating the AUC score
from sklearn.metrics import roc_auc_score

auc_logistic = roc_auc_score(y_test_ohe, y_probability_Logistic)
auc_svm = roc_auc_score(y_test_ohe, y_probability_svm)
auc_rf = roc_auc_score(y_test_label, y_probability_random_forest)
auc_xgb = roc_auc_score(y_test_label, y_probability_xgb_label)

print(f"AUC - Logistic Regression: {auc_logistic:.4f}")
print(f"AUC - SVM: {auc_svm:.4f}")
print(f"AUC - Random Forest: {auc_rf:.4f}")
print(f"AUC - XGBoost: {auc_xgb:.4f}")
```

```python
#Plotting the ROC Curve

#Logistic Regression (OHE)
fpr_logistic, tpr_logistic, _ = roc_curve(y_test_ohe, y_probability_Logistic)

#Random Forest Classifier
fpr_rf, tpr_rf, _ = roc_curve(y_test_label, y_probability_random_forest)

#Xg Boost
fpr_xg, tpr_xg, _ = roc_curve(y_test_label, y_probability_xgb_label)

#SVM(OHE)
fpr_svm, tpr_svm , _ = roc_curve(y_test_ohe, y_probability_svm)

#PLoting all in one graph
plt.figure(figsize = (8,6))
plt.plot(fpr_logistic, tpr_logistic, label = f'Logistic Regression (AUC = {auc_logistic:.4f})')
plt.plot(fpr_svm, tpr_svm, label = f'SVM (AUC = {auc_svm:.4f})')
plt.plot(fpr_rf, tpr_rf, label = f'Random Forest(AUC = {auc_rf:.4f})')
plt.plot(fpr_xg, tpr_xg, color='orange', label=f'XGBoost (Label) AUC = {auc_xgb:.4f}')


plt.plot([0,1], [0,1], 'k--', label = 'Random')

plt.xlabel ('False Positive Rate')
plt.ylabel ('True Positive Rate')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

Random Forest performs the best with an AUC of 0.9873.  
XGBoost comes very close, with an AUC of 0.9868  
SVM also performs very well with an AUC of 0.9667  
Logistic Regression has the lowest AUC among the three (0.9525) but still shows strong performance.

<span style="font-size:30px">
Model Fine-Tuning

```python
## Model Tuning with GridSearchCV
from sklearn.model_selection import GridSearchCV
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [5, 10, None],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}

rf = RandomForestClassifier(random_state = 42)
grid_search = GridSearchCV(estimator = rf, param_grid = param_grid, cv = 5, scoring = 'f1', n_jobs = -1, verbose = 1)
grid_search.fit(x_train_label, y_train_label)

best_rf = grid_search.best_estimator_
y_predict_best = best_rf. predict(x_test_label)
y_probability_best = best_rf.predict_proba(x_test_label)[:,1]

#Evaluate
evaluate_model("Tuned Random Forest", y_test_label, y_predict_best, y_probability_best)
print("Best parameters:", grid_search.best_params_)
```

There's a slight improvement after applying tunning through GridSearchCV.  
The tuned model achieved:  
Accuracy: 0.9488 
F1 Score: 0.9125  
AUC-ROC: 0.9873  

Earlier it was:
Accuracy:   0.9467  
F1-score:   0.9069  
AUC-ROC:    0.9842


```python
#Cross-Validation for Best Performing Model which is Random Forest
from sklearn.model_selection import cross_validate
scoring = ['accuracy', 'precision', 'recall', 'f1', 'roc_auc']
cv_scores = cross_validate(estimator = best_rf, X = x_train_label, y = y_train_label, cv = 5, scoring =scoring, return_train_score=False)

print("5-Fold Cross-Validation Results (Tuned Random Forest):\n")
print("Accuracy scores:", cv_scores['test_accuracy'])
print("Average Accuracy:", cv_scores['test_accuracy'].mean())

print("\nPrecision scores:", cv_scores['test_precision'])
print("Average Precision:", cv_scores['test_precision'].mean())

print("\nRecall scores:", cv_scores['test_recall'])
print("Average Recall:", cv_scores['test_recall'].mean())

print("\nF1 scores:", cv_scores['test_f1'])
print("Average F1 Score:", cv_scores['test_f1'].mean())

print("\nROC AUC scores:", cv_scores['test_roc_auc'])
print("Average ROC AUC:", cv_scores['test_roc_auc'].mean())
```

The tuned Random Forest model shows high and consistent performance across all folds.  

These results show that the model is robust, generalizes well, and is reliable for predicting app success based on the features used.

```python
#Voting Classifier
from sklearn.ensemble import VotingClassifier
voting_classifier = VotingClassifier (estimators = [
    ('lr', Logistic),
    ('svm', svm),
    ('rf', RandomForestClassifier(random_state = 42))],
    voting = 'hard'
)

#train and predicting the model using OHE data for Logistic, SVM, RandomForest
voting_classifier.fit(x_train_ohe, y_train_ohe)
y_pred_voting = voting_classifier.predict(x_test_ohe)

#Accuracy and AUC
voting_acc = accuracy_score(y_test_ohe, y_pred_voting)
voting_auc = roc_auc_score(y_test_ohe, y_pred_voting)

print("Voting Classifier Accuracy:", voting_acc)
print("Voting Classifier AUC:", voting_auc)
```

```python
#Stacking Classifier
from sklearn.ensemble import StackingClassifier
stacking_classifier = StackingClassifier(
    estimators = [
        ('lr', Logistic),
        ('svm', svm),
        ('rf', RandomForestClassifier(random_state = 42))],
        final_estimator = LogisticRegression()
)

stacking_classifier.fit(x_train_ohe, y_train_ohe)
y_predict_stacking = stacking_classifier.predict(x_test_ohe)
y_probability_stacking = stacking_classifier.predict_proba(x_test_ohe)[:,1]

stacking_accuracy = accuracy_score(y_test_ohe, y_predict_stacking)
stacking_auc = roc_auc_score(y_test_ohe, y_probability_stacking)

print("Stacking Classifier Accuracy:", stacking_accuracy)
print("Stacking Classifier AUC:", stacking_auc)
```
